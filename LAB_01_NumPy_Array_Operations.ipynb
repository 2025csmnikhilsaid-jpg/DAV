```python
print("Third Element:",arr[2])
```

```python
print("Last three elements:",arr[-3:])
```

```python
import pandas as pd
```

```python
data={
    "Name":["Asha","Ravi","Meera","Karan"],
    "Marks":[88,72,91,65],
    "Branch":["CSE","ECE","CSE","ISE"]
}
```

```python
df=pd.DataFrame(data)
```

```python
print("\n---Pandas DataFrame---")
```

```python
print(df)
```

```python
print("\n---Column Indexing---")
```

```python
print("Name Column:")
```

```python
print(df["Name"])
```

```python
print("\nMarks Column:")
```

```python
print(df["Marks"])
```

```python
print("\n---Row Indexing---")
```

```python
print("First Row:")
```

```python
print(df.iloc[0])
```

```python
print("\nThird Row:")
```

```python
print(df.iloc[2])
```

```python
print("\n---Specific Value---")
```

```python
print("Marks of first student:",df.iloc[0,1])
```

```python
print("\n---Row Slicing---")
```

```python
print("First two rows:")
```

```python
print(df.iloc[0:2])
```

```python
print("First three rows with Names and Marks:")
```

```python
print(df.iloc[0:3,[0,1]])
```

```python
print("\n---Data Filtering---")
```

```python
result=df[df["Marks"]>70]
```

```python
print("Students with Marks Greater thsn 70:")
```

```python
print(result)
```
