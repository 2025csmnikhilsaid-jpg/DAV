```python
import pandas as pd
import numpy as np
```

```python
data ={
    "Student" : ["A","B","C","D","E",
                 "F","G","H","I","J"],
    "Marks" : [50,55,60,62,65,68,70,72,75,150]
}
```

```python
df = pd.DataFrame(data)
```

```python
print("Original Dataset:")
print(df)
```

```python
Q1 = df["Marks"].quantile(0.25)
print(Q1)
```

```python
Q3 = df["Marks"].quantile(0.75)
print(Q3)
```

```python
IQR = Q3-Q1
print(IQR)
```

```python
lower_limit = Q1-1.5*IQR
upper_limit = Q3+1.5*IQR
```

```python
print("\nQ1",Q1)
print("\nQ3",Q3)
print("\nIQR",IQR)
print("\nLower Limit ",lower_limit)
print("\nUpper Limit ",upper_limit)
```

```python
outliers = df[(df["Marks"]< lower_limit)|
              (df["Marks"]> upper_limit)]
```

```python
print("\nOutlier :",outliers)
```

```python
df["Marks_Treated"] = np.where(
    df["Marks"]>upper_limit,
    upper_limit,
    np.where(
        df["Marks"]< lower_limit,
        lower_limit,
        df["Marks"]
    )

)
```

```python
# print("\nOutliers using np")
df_capped = df.copy()
df_capped["Marks"] = df_capped["Marks"].clip(
    lower = lower_limit,
    upper = upper_limit
)
print("\nAfter Capping:")
print(df_capped)
```

```python
print("\nDataSet After Outlier Treatment:")
print(df)
```
